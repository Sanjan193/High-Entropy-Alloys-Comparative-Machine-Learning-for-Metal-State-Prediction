# **Dropping the unnecessary features**

```python
# dropping the text featres here to get the accurate numeric data rame
data.drop(columns=['Alloy ID','Quenching','HPR','Unnamed: 51','Annealing_Temp','Unnamed: 52','References','Hot-Cold_Working','IM_Structure','Alloy ','Microstructure_', 'Multiphase','Microstructure'],axis=1,inplace=True)
```

To get the more refined data set I'm here dropping all the rows except the **"Annealing_Temp"** 603 values

"Sythesis_Route","Hot-Cold_Working","Annealing_Temp","Microstructure_","Multiphase","IM_Structure","Microstructure","Phases"

```python
# Final data set till here
df_final = data[data["Annealing_Time_(min)"].notnull()]
```

# **Fillling Null Values And Trnsforming The Text Data Into Embiddings**

In the data set 2 columns **'Homogenization_Temp', 'Homogenization_Time'** has only somme null value....

So I'm going to fill the 'Homogenization_Time' as the mode vale and the 'Homogenization_Temp' as the

```python
df_final['Homogenization_Temp'].fillna(df_final['Homogenization_Temp'].median(),inplace=True)
df_final['Homogenization_Time'].fillna(df_final['Homogenization_Time'].mode().iloc[0],inplace=True)
```

# **Encoding the text data into numbers**

```python
# for Sythesis Route
map1 = {
   'AC':1,
   'PM':2
}
# for phases
map2 = {
    'BCC_SS':1,
    'FCC_SS':2,
    'Im':3,
    'FCC_PLUS_BCC':4
}

df_final['Sythesis_Route'] = df_final['Sythesis_Route'].map(map1)
df_final['Phases'] = df_final['Phases'].map(map2)
```
